# Module 1: Element-wise tensor operations

```python
import torch
```

```python
m1 = torch.tensor([
    [1, 2, 4],
    [-1, 3, -1],
    [0, 0, 2]
])
m2 = torch.tensor([
    [0, 2, 2],
    [-2, 2, 0],
    [1, 1, 1]
])
```

```python
m1 + m2
```

```python
m1 - m2
```

```python
m1 / m2
```

```python
m1 * m2
```

```python
3 * m1
```

```python
torch.tensor(3) * m1
```

```python
3 == torch.tensor(3)
```
