# Module 1 first tensors lecture

```python
import torch
from IPython.display import display

print(torch.__version__)
print(torch.cuda.is_available())
```

```python
x = torch.tensor(7)
display(x)
display(x.shape)
```

```python
v_1 = torch.tensor([-1, 99, 5, 19])
display(v_1)
display(v_1.shape)
```

```python
m_1 = torch.tensor([
    [5, -1, 0],
    [2, 0, 0]
])
display(m_1)
display(m_1.shape)
```

```python
s_1 = torch.tensor([
    [
        [1, 0],
        [0, 1]
    ],
    [
        [0, 0],
        [1, 2]
    ],
    [
        [1, 1],
        [1, 1]
    ]
])
display(s_1)
display(s_1.shape)
```
