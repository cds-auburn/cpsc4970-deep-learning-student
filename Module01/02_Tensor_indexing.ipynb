# Module 1: Tensor indexing

```python
import torch
```

```python
v = torch.tensor([1, -1, 0, 12, 22])
m = torch.tensor([
    [1, 5, 10],
    [2, 19, -5],
    [1, 1, 55],
    [-3, 18, -1]
])
print(f"{v=}, {v.shape=}")
print(f"{m=}, {m.shape=}")
```

Accessing individual elements

```python
v[0]
```

```python
v[3]
```

```python
m[0, 2]
```

```python
m[3, 1]
```

Slicing and dicing

```python
v[1:3]
```

```python
m[0]
```

```python
m[2]
```

```python
m[1:3]
```

```python
m[..., 0]
```

```python
m[..., 1:3]
```

start:stop:step

```python
print(f"{v=}")
v[1:5:3]
```

```python
v[:2]
```

```python
v[::2]
```

```python
v[:]
```

```python
print(f"{m=}")
m[:, 1]
```

Indexing from the end

```python
v[-1]
```

```python
m[-1]
```

```python
m[:, -1]
```

Adding/removing dimensions

```python
print(f"{v=}, {v.shape=}")
z = v[None, :]
print(f"{z=}, {z.shape=}")
```

```python
q = v[:, None]
print(f"{q=}, {q.shape=}")
```

```python
q.squeeze(1)
```

```python
z.squeeze(0)
```

Creating flag tensor and using as indices

```python
print(f"{v=}")
v[[False, True, True, True, False]]
```

```python
v > 0
```

```python
v[v > 0]
```

```python
v == 12
```
